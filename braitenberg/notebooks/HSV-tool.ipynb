```python
# Import necessary libraries
import cv2
import numpy as np
from matplotlib import pyplot as plt
from ipywidgets import interact, IntSlider, VBox

# Load the image
image_path = "/code/braitenberg/assets/samples/duckie-images/image2.png"  # Change this to the path of your image
image = cv2.imread(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
image_hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)

# Function to apply and display HSV range filtering
def apply_hsv_filter(low_h, high_h, low_s, high_s, low_v, high_v):
    # Define the lower and upper HSV bounds
    lower_bound = np.array([low_h, low_s, low_v])
    upper_bound = np.array([high_h, high_s, high_v])
    
    # Apply the HSV range mask
    mask = cv2.inRange(image_hsv, lower_bound, upper_bound)
    result = cv2.bitwise_and(image_rgb, image_rgb, mask=mask)

    # Display the original and filtered images
    plt.figure(figsize=(12, 6))
    plt.subplot(1, 2, 1)
    plt.title("Original Image")
    plt.imshow(image_rgb)
    plt.axis("off")
    
    plt.subplot(1, 2, 2)
    plt.title("Filtered Image")
    plt.imshow(result)
    plt.axis("off")
    plt.show()

# Create sliders for low and high HSV values
interact(apply_hsv_filter,
         low_h=IntSlider(min=0, max=179, step=1, value=0, description='Low H'),
         high_h=IntSlider(min=0, max=179, step=1, value=179, description='High H'),
         low_s=IntSlider(min=0, max=255, step=1, value=0, description='Low S'),
         high_s=IntSlider(min=0, max=255, step=1, value=255, description='High S'),
         low_v=IntSlider(min=0, max=255, step=1, value=0, description='Low V'),
         high_v=IntSlider(min=0, max=255, step=1, value=255, description='High V'))
```
